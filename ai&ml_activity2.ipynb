```python

# ============================================================
# 🐾 ANIMALVISION AI - COMPLETE GOOGLE COLAB PROJECT
# Modern AI Dashboard | Gradio | Hugging Face CLIP
# ============================================================

!pip install -q gradio transformers torch pillow pandas

import gradio as gr
import torch
import pandas as pd
import tempfile
import html
from datetime import datetime
from transformers import pipeline

# ============================================================
# 1. ANIMAL CATEGORIES
# ============================================================

ANIMALS = [
    "dog", "cat", "lion", "tiger", "bear",
    "elephant", "horse", "cow", "rabbit", "bird",
    "parrot", "penguin", "giraffe", "zebra", "monkey",
    "deer", "fox", "wolf", "panda", "crocodile",
    "snake", "turtle", "frog", "fish", "shark",
    "dolphin", "whale", "chicken", "duck", "sheep",
    "goat", "pig", "hamster", "squirrel", "kangaroo",
    "camel", "owl", "eagle", "peacock", "butterfly"
]

LABELS = [f"a photo of a {animal}" for animal in ANIMALS]

# ============================================================
# 2. ANIMAL FACTS
# ============================================================

FACTS = {
    "dog": "Dogs are loyal domestic animals with an excellent sense of smell.",
    "cat": "Cats are independent mammals with excellent night vision.",
    "lion": "Lions are social big cats that live in groups called prides.",
    "tiger": "Tigers are powerful striped cats and excellent swimmers.",
    "bear": "Bears are large mammals with a strong sense of smell.",
    "elephant": "Elephants have remarkable memories and long trunks.",
    "horse": "Horses are strong and fast mammals.",
    "cow": "Cows are herbivorous mammals commonly raised on farms.",
    "rabbit": "Rabbits are herbivorous mammals with powerful hind legs.",
    "bird": "Birds are warm-blooded animals with feathers and beaks.",
    "parrot": "Parrots are colorful birds known for intelligence.",
    "penguin": "Penguins are flightless birds and excellent swimmers.",
    "giraffe": "Giraffes are the tallest land animals.",
    "zebra": "Zebras are known for their black and white stripes.",
    "monkey": "Monkeys are intelligent primates.",
    "deer": "Deer are herbivorous mammals; many species grow antlers.",
    "fox": "Foxes are clever mammals with excellent hearing.",
    "wolf": "Wolves are social wild canines.",
    "panda": "Giant pandas mainly eat bamboo.",
    "crocodile": "Crocodiles are large reptiles that live near water.",
    "snake": "Snakes are legless reptiles.",
    "turtle": "Turtles are reptiles protected by a hard shell.",
    "frog": "Frogs are amphibians that often live near water.",
    "fish": "Fish are aquatic animals that generally breathe through gills.",
    "shark": "Sharks are marine fish with cartilaginous skeletons.",
    "dolphin": "Dolphins are intelligent marine mammals.",
    "whale": "Whales are large marine mammals that breathe air.",
    "chicken": "Chickens are domesticated birds.",
    "duck": "Ducks are waterbirds with webbed feet.",
    "sheep": "Sheep are domesticated mammals commonly raised for wool.",
    "goat": "Goats are adaptable herbivorous mammals.",
    "pig": "Pigs are intelligent omnivorous mammals.",
    "hamster": "Hamsters are small rodents.",
    "squirrel": "Squirrels are rodents known for collecting nuts.",
    "kangaroo": "Kangaroos are Australian marsupials.",
    "camel": "Camels are desert-adapted mammals.",
    "owl": "Owls are birds of prey with excellent night vision.",
    "eagle": "Eagles are powerful birds of prey.",
    "peacock": "Peacocks are famous for colorful feathers.",
    "butterfly": "Butterflies are insects with colorful wings."
}

# ============================================================
# 3. LOAD AI MODEL
# ============================================================

print("Loading AnimalVision AI...")

device = 0 if torch.cuda.is_available() else -1

print("Running on:", "GPU" if device == 0 else "CPU")

classifier = pipeline(
    "zero-shot-image-classification",
    model="openai/clip-vit-base-patch32",
    device=device
)

print("AI MODEL LOADED SUCCESSFULLY!")

# ============================================================
# 4. DASHBOARD CSS
# ============================================================


CSS = """
@import url('https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700;800&display=swap');

.gradio-container {
    font-family: Inter, sans-serif !important;
    background: #f1f5f9 !important;
    color: #172554 !important;
    max-width: 1250px !important;
}

/* Main headings and text */
.gradio-container h1,
.gradio-container h2,
.gradio-container h3,
.gradio-container h4,
.gradio-container p,
.gradio-container label,
.gradio-container span {
    color: #172554;
}

/* Dashboard header */
#header {
    background: linear-gradient(120deg,#0f172a,#1e40af,#6d28d9);
    padding: 30px;
    border-radius: 18px;
    margin-bottom: 20px;
    color: #ffffff !important;
}

#header h1 {
    color: #ffffff !important;
    font-size: 34px;
    font-weight: 800;
}

#header p {
    color: #e0e7ff !important;
}

/* Statistics cards */
.stat {
    background: #ffffff;
    padding: 18px;
    border-radius: 14px;
    border-left: 4px solid #3b82f6;
    box-shadow: 0 4px 12px #00000010;
}

.stat h2 {
    color: #1d4ed8 !important;
}

.stat p, .stat small {
    color: #475569 !important;
}

/* Result card */
.result {
    background: #ffffff;
    padding: 24px;
    border-radius: 16px;
    border-top: 5px solid #4f46e5;
    box-shadow: 0 5px 20px #00000012;
    color: #172554 !important;
}

.result h1 {
    color: #1d4ed8 !important;
    text-align: center;
}

.result h2, .result h3 {
    color: #1e3a8a !important;
}

.animal {
    font-size: 36px;
    text-align: center;
}

.score {
    color: #047857 !important;
    font-size: 38px;
    font-weight: 800;
    text-align: center;
}

.muted {
    color: #475569 !important;
    text-align: center;
}

/* Animal facts */
.fact {
    background: #eff6ff;
    color: #1e3a8a !important;
    padding: 15px;
    border-radius: 10px;
    line-height: 1.7;
}

/* Prediction chart */
.chart {
    background: #ffffff;
    padding: 20px;
    border-radius: 15px;
    color: #172554 !important;
}

.chart h3 {
    color: #1e3a8a !important;
}

.bar-row {
    margin: 17px 0;
}

.bar-label {
    display: flex;
    justify-content: space-between;
    font-size: 13px;
    margin-bottom: 7px;
    color: #172554 !important;
}

.bar-label span,
.bar-label b {
    color: #172554 !important;
}

.track {
    height: 12px;
    background: #cbd5e1;
    border-radius: 20px;
    overflow: hidden;
}

.fill {
    height: 100%;
    background: linear-gradient(90deg,#2563eb,#8b5cf6);
    border-radius: 20px;
    animation: grow 1s ease;
}

@keyframes grow {
    from {width:0;}
}

/* History */
.history {
    background: #ffffff;
    padding: 20px;
    border-radius: 15px;
    color: #172554 !important;
}

.history h3 {
    color: #1e3a8a !important;
}

.history-row {
    display: flex;
    justify-content: space-between;
    padding: 12px;
    border-bottom: 1px solid #e2e8f0;
    font-size: 12px;
    color: #172554 !important;
}

.history-row span,
.history-row small {
    color: #334155 !important;
}

.history-row b {
    color: #1d4ed8 !important;
}

/* Input labels */
.gradio-container label {
    color: #1e293b !important;
    font-weight: 600 !important;
}

/

/* DATA TABLE - TOP 5 PREDICTIONS */

/* Table container */
.gradio-container .table-wrap,
.gradio-container .dataframe,
.gradio-container [role="grid"] {
    background-color: #ffffff !important;
    color: #111827 !important;
}

/* Table headers */
.gradio-container th,
.gradio-container thead th,
.gradio-container [role="columnheader"] {
    background-color: #1e3a8a !important;
    color: #ffffff !important;
    font-weight: 800 !important;
}

/* Header text */
.gradio-container th *,
.gradio-container [role="columnheader"] * {
    color: #ffffff !important;
}

/* Table cells */
.gradio-container td,
.gradio-container tbody td,
.gradio-container [role="gridcell"] {
    background-color: #ffffff !important;
    color: #111827 !important;
    font-weight: 600 !important;
}

/* Cell text */
.gradio-container td *,
.gradio-container [role="gridcell"] * {
    color: #111827 !important;
}

/* Alternate rows */
.gradio-container tbody tr:nth-child(even) td {
    background-color: #eff6ff !important;
    color: #111827 !important;
}

/* Table borders */
.gradio-container table,
.gradio-container th,
.gradio-container td {
    border: 1px solid #cbd5e1 !important;
}

/* Buttons */
#analyze {
    background: linear-gradient(100deg,#2563eb,#7c3aed) !important;
    color: white !important;
    font-weight: 700 !important;
    border: none !important;
}

#analyze span {
    color: white !important;
}

#reset-button {
    border-radius: 10px !important;
}

/* Footer */
footer {
    display: none !important;
}
"""



# ============================================================
# 5. RECOGNITION FUNCTION
# ============================================================

def recognize(image, history):

    if image is None:
        return (
            "<div class='result'>Please upload an image!</div>",
            "",
            pd.DataFrame(columns=["Rank", "Animal", "Score"]),
            "<p>No history available.</p>",
            history,
            None
        )

    image = image.convert("RGB")

    predictions = classifier(
        image,
        candidate_labels=LABELS,
        hypothesis_template="This is {}."
    )

    best = predictions[0]

    animal = best["label"].replace("a photo of a ", "").title()
    score = float(best["score"]) * 100

    facts = FACTS.get(
        animal.lower(),
        "Explore more about this animal."
    )

    # Result card
    result_html = f"""
    <div class="result">
        <div class="animal">🐾</div>
        <p class="muted">PREDICTED ANIMAL</p>
        <h1>{html.escape(animal.upper())}</h1>
        <div class="score">{score:.2f}%</div>
        <p class="muted">Similarity Score</p>
        <hr>
        <h3>🌍 Animal Information</h3>
        <div class="fact">{html.escape(facts)}</div>
        <br>
        <small class="muted">
        Similarity scores are not calibrated probabilities.
        </small>
    </div>
    """

    # Top 5 chart and table
    chart = "<div class='chart'><h3>📊 Top 5 Predictions</h3>"

    rows = []

    for i, item in enumerate(predictions[:5], 1):

        name = item["label"].replace(
            "a photo of a ", ""
        ).title()

        percentage = float(item["score"]) * 100

        chart += f"""
        <div class="bar-row">
            <div class="bar-label">
                <span>{i}. {html.escape(name)}</span>
                <b>{percentage:.2f}%</b>
            </div>
            <div class="track">
                <div class="fill" style="width:{percentage:.2f}%"></div>
            </div>
        </div>
        """

        rows.append([i, name, f"{percentage:.2f}%"])

    chart += "</div>"

    table = pd.DataFrame(
        rows,
        columns=["Rank", "Animal", "Score"]
    )

    # History
    new_history = list(history or [])

    new_history.append({
        "Date": datetime.now().strftime("%d-%m-%Y %H:%M"),
        "Animal": animal,
        "Score": f"{score:.2f}%"
    })

    history_html = "<div class='history'><h3>📚 Recognition History</h3>"

    for item in reversed(new_history[-10:]):

        history_html += f"""
        <div class="history-row">
            <span>🐾 {html.escape(item['Animal'])}</span>
            <b>{item['Score']}</b>
            <small>{item['Date']}</small>
        </div>
        """

    history_html += "</div>"

    # Downloadable report
    report = f"""
ANIMALVISION AI REPORT
=============================

Date: {datetime.now()}

Predicted Animal: {animal}
Similarity Score: {score:.2f}%

TOP 5 PREDICTIONS
"""

    for row in rows:
        report += f"{row[0]}. {row[1]} - {row[2]}\n"

    report += f"\nANIMAL FACTS\n{facts}\n"

    file = tempfile.NamedTemporaryFile(
        mode="w",
        suffix=".txt",
        delete=False,
        encoding="utf-8"
    )

    file.write(report)
    file.close()

    return (
        result_html,
        chart,
        table,
        history_html,
        new_history,
        file.name
    )

# ============================================================
# 6. RESET FUNCTION
# ============================================================

def reset():

    return (
        None,
        "<div class='result'><h2>🐾 Ready for Analysis</h2></div>",
        "",
        pd.DataFrame(columns=["Rank", "Animal", "Score"]),
        "<p>No history available.</p>",
        [],
        None
    )

# ============================================================
# 7. BUILD INTERACTIVE DASHBOARD
# ============================================================

with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="violet"
    ),
    css=CSS,
    title="AnimalVision AI"
) as demo:

    history_state = gr.State([])

    gr.HTML("""
    <div id="header">
        <h1>🐾 AnimalVision AI</h1>
        <p>Modern Artificial Intelligence Animal Recognition Dashboard</p>
        <p>Powered by Hugging Face CLIP</p>
        <span>🟢 AI SYSTEM ONLINE</span>
    </div>
    """)

    with gr.Row():

        gr.HTML("""
        <div class="stat">
            <small>AI MODEL</small>
            <h2>CLIP</h2>
            <p>Vision AI</p>
        </div>
        """)

        gr.HTML("""
        <div class="stat">
            <small>ANIMAL CATEGORIES</small>
            <h2>40</h2>
            <p>Supported classes</p>
        </div>
        """)

        gr.HTML("""
        <div class="stat">
            <small>PREDICTIONS</small>
            <h2>TOP 5</h2>
            <p>Ranked results</p>
        </div>
        """)

    gr.Markdown("## 🔍 Animal Recognition")

    with gr.Row():

        with gr.Column():

            image_input = gr.Image(
                label="Upload Animal Image",
                type="pil",
                height=350
            )

            analyze_btn = gr.Button(
                "🔍 ANALYZE IMAGE",
                variant="primary",
                elem_id="analyze"
            )

            reset_btn = gr.Button("↻ RESET DASHBOARD")

            gr.Markdown(
                "Supported formats: JPG, JPEG, PNG, WEBP"
            )

        with gr.Column():

            result_output = gr.HTML(
                "<div class='result'><h2>🐾 Ready for Analysis</h2><p>Upload an image to begin.</p></div>"
            )

    gr.Markdown("---")
    gr.Markdown("## 📊 Prediction Analytics")

    chart_output = gr.HTML()

    prediction_table = gr.Dataframe(
        headers=["Rank", "Animal", "Score"],
        interactive=False,
        label="Top 5 Predictions"
    )

    gr.Markdown("---")
    gr.Markdown("## 📚 Recognition History")

    history_output = gr.HTML(
        "<p>No history available.</p>"
    )

    gr.Markdown("---")
    gr.Markdown("## 📥 Download Report")

    report_output = gr.File(
        label="Download Recognition Report"
    )

    gr.HTML("""
    <div style="
        background:#0f172a;
        color:white;
        padding:25px;
        text-align:center;
        border-radius:12px;
        margin-top:20px;">
        <h3>🐾 ANIMALVISION AI</h3>
        Python | Gradio | Hugging Face | CLIP
        <br><br>

    </div>
    """)

    # Analyze button
    analyze_btn.click(
        fn=recognize,
        inputs=[image_input, history_state],
        outputs=[
            result_output,
            chart_output,
            prediction_table,
            history_output,
            history_state,
            report_output
        ]
    )

    # Reset button
    reset_btn.click(
        fn=reset,
        inputs=[],
        outputs=[
            image_input,
            result_output,
            chart_output,
            prediction_table,
            history_output,
            history_state,
            report_output
        ]
    )

# ============================================================
# 8. LAUNCH APP
# ============================================================

print("Launching AnimalVision AI...")

demo.queue()

demo.launch(
    share=True,
    debug=False,
    show_error=True
)

# ============================================================
# PROJECT COMPLETED
# ============================================================
```
